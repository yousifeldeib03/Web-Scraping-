```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
import pandas as pd

driver = webdriver.Edge()
driver.get("https://www.imdb.com/chart/toptv/")

time.sleep(5)
list_movies = []
elements = driver.find_elements(By.CSS_SELECTOR, "li.ipc-metadata-list-summary-item")
for element in elements:
  name = element.find_element(By.CSS_SELECTOR, "h4.ipc-title__text")
  lot = element.find_element(By.CLASS_NAME, "cli-title-metadata")
  year = lot.find_element(By.CSS_SELECTOR, " li:nth-child(1)")
  type = lot.find_element(By.CSS_SELECTOR, " li:nth-child(2)")
  rate = element.find_element(By.CLASS_NAME, "ipc-rating-star--rating")
  ranking = element.find_element(By.CLASS_NAME, "ipc-signpost__text")
  list_movies.append({
    "ranking": ranking.text,
    "name": name.text,
    "year": year.text,
    "type": type.text,
    "rate": rate.text
  })

    

driver.quit()
df= pd.DataFrame(list_movies)
df.to_csv("tv_shows.csv")

    
```
