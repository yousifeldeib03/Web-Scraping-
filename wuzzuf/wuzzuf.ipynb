```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
import pandas as pd
list_of_jobs = []
driver = webdriver.Edge()
driver.maximize_window()
driver.get("https://wuzzuf.net/jobs/egypt")
driver.find_element(By.NAME, "q").send_keys("Data Engineer")
driver.find_element(By.XPATH, "//*[@id='app']/div/div/main/div[2]/div[1]/form/button").click()
time.sleep(5)
for i in range(40):
  elements = driver.find_elements(By.CSS_SELECTOR, "div.css-ghe2tq.e1v1l3u10")
  for element in elements:
    title = element.find_element(By.CLASS_NAME , "css-o171kl").text
    company = element.find_element(By.CLASS_NAME , "css-ipsyv7").text
    location = element.find_element(By.CLASS_NAME , "css-16x61xq").text
    type = element.find_element(By.CLASS_NAME , "css-5jhz9n").text
    list_of_jobs.append({"Title": title, "Company": company, "Location": location, "Type": type})
  buttons = driver.find_elements(By.CSS_SELECTOR, "button.css-kefogq.ezfki8j0")
  buttons[-1].click()
  time.sleep(5)

df = pd.DataFrame(list_of_jobs)
df.to_csv("wuzzuf_jobs.csv", index=False)
driver.quit()





    
```
