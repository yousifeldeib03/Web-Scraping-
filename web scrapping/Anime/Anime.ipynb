```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
import pandas as pd

driver = webdriver.Edge()
driver.get("https://www.imdb.com/list/ls057577566")

time.sleep(5)
list_animes = []
elements = driver.find_elements(By.CSS_SELECTOR, "li.ipc-metadata-list-summary-item")
for element in elements:
  name = element.find_element(By.CSS_SELECTOR, "h4.ipc-title__text")
  lot = element.find_element(By.CLASS_NAME, "dli-title-metadata")
  year = lot.find_element(By.CSS_SELECTOR, " li:nth-child(1)")
  year_text = year.text
  year_text = year_text.replace("â€“", "–")
  if "–" in year_text:
    start, end = year_text.split("–")
  else:
    start = end = year_text
  type = lot.find_element(By.CSS_SELECTOR, " li:nth-child(2)")
  rate = element.find_element(By.CLASS_NAME, "ipc-rating-star--rating")
  votes = element.find_element(By.CLASS_NAME, "ipc-rating-star--voteCount")
  
  list_animes.append({
    "name": name.text,
    "start": start,
    "end": end,
    "type": type.text,
    "rate": rate.text ,
    "votes": votes.text
  })

    

driver.quit()
df= pd.DataFrame(list_animes)
df.to_csv("anime.csv",index=False)
```
